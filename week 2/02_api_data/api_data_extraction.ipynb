## Install /use the API library

```python
import requests
import pandas as pd
```

## call API...it shows the GET HTTP method

```python
url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

print("Status Code:", response.status_code)
```

## Inspect the JSON response

```python
data= response.json()
print(data)
```

## Convert JSON into a Pandas DataFrame

```python
df_api=pd.DataFrame(data)
df_api.head()
```

## save as CSV

```python
df_api.to_csv("api_users.csv", index=False)
print("CSV file created successfully")
```

## Basic Analysis

```python
print("Total Records:",len(df_api))
print("Total Columns:",len(df_api.columns))
print(df_api.columns.tolist())
```

## Conclusion

- Successfully extracted data from a public REST API using the GET method.
- Inspected the JSON response returned by the API.
- Converted JSON data into a Pandas DataFrame.
- Exported the extracted data into CSV format.
- Verified the CSV file and prepared the data for further analysis.
